## QUESTION 4

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])
```

```python
print(X.shape)
print(X.ndim)
```

```python
X_trans = X.T
X_trans
```

```python
XTX = X_trans @ X
XTX
```

```python
inverse_XTX = np.linalg.inv(XTX)
inverse_XTX
```

```python
beta = inverse_XTX @ X_trans @ y
beta
```

```python
new_user = np.array([5, 40, 7])

prediction = new_user @ beta
prediction
```
