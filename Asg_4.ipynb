## QUESTION 1


```python
import pandas as pd


roll_number = input("Enter your college roll number: ")

last_two_digits = [int(digit) for digit in roll_number[-2:]]


categories = ["billing", "account", "general"]



fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },

    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },

    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },

    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]



personalized_entries = []

for digit in last_two_digits:

   
    category = categories[digit % 3]

    # Create an FAQ according to category
    if category == "billing":
        question = "how can i check my payment status"
        answer = "You can check your payment status from the billing section."
        keywords = "payment transaction status"

    elif category == "account":
        question = "how do i update my registered mobile number"
        answer = "Go to Account Settings and update your registered mobile number."
        keywords = "mobile number update account"

    else:
        question = "how can i contact customer support"
        answer = "You can contact customer support through the help section."
        keywords = "support help contact"

    personalized_entries.append({
        "question": question,
        "answer": answer,
        "keywords": keywords,
        "category": category
    })




all_entries = fixed_entries + personalized_entries

df = pd.DataFrame(all_entries)



print("\nFinal 6-Row FAQ DataFrame:\n")
df
```

## QUESTION 2

```python
def score_query(query, df):

    # Convert query into lowercase words
    query_words = query.lower().split()

    results = []

    # Check every FAQ entry
    for index, row in df.iterrows():

        # Combine question and keywords
        text = (row["question"] + " " + row["keywords"]).lower()

        # Count matching words
        score = sum(word in text for word in query_words)

        # Add only matching entries
        if score > 0:
            confidence = score / len(query_words)

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score,
                "confidence": confidence
            })

    # Convert results into DataFrame
    result_df = pd.DataFrame(results)

    # Sort by highest confidence
    if not result_df.empty:
        result_df = result_df.sort_values(
            by=["confidence", "score"],
            ascending=False
        )

    return result_df
```

```python
score_query("fee payment", df)
```

## QUESTION 3

```python
# Q3: Find all FAQs belonging to a category

def same_category(category_name, df):

    return df[df["category"] == category_name]


# Get category from the first personalized entry
personalized_category = personalized_entries[0]["category"]

print("Selected category:", personalized_category)


# Call the function
category_result = same_category(
    personalized_category,
    df
)


# Print the result
print("\nFAQs belonging to this category:")
category_result
```

## QUESTION 4

```python
# Display all FAQ entries
print("\nAvailable FAQ entries:")

for i in range(len(df)):
    print(i, ":", df.loc[i, "question"])


# Ask user which entry to update
choice = int(
    input("\nEnter the entry number to update: ")
)


# Ask for a new keyword
new_keyword = input(
    "Enter a new keyword: "
).strip().lower()


# Add the new keyword
df.loc[choice, "keywords"] = (
    df.loc[choice, "keywords"]
    + " "
    + new_keyword
)


# Create CSV filename
filename = roll_number + "_faq_data.csv"


# Save the entire DataFrame
df.to_csv(
    filename,
    index=False
)


# Display updated DataFrame
print("\nUpdated DataFrame:")
print(df)

print("\nData saved successfully as:", filename)
```
