```python
roll_no = input("Enter your 10-digit roll number: ")

# Convert each digit into an integer and multiply by 10
digits = [int(digit) for digit in roll_no]
L = [digit * 10 for digit in digits]

# i. Print L
print("\ni. Original List L:")
print(L)
```

```python
L.append(250)
print("\nAfter append(250):")
print(L)
# append() adds the new element at the END of the list.

L.insert(2, 350)
print("\nAfter insert(2, 350):")
print(L)
```

```python
L.remove(250)
print("\nAfter remove(250):")
print(L)
# remove() deletes the FIRST occurrence of the specified value.

removed_element = L.pop()
print("\nAfter pop():")
print(L)
print("Element removed using pop():", removed_element)
```

```python
L.sort()
print("\nAscending order:")
print(L)

L.sort(reverse=True)
print("\nDescending order:")
print(L)
```

```python
print("\nFirst three elements:", L[:3])
print("Last three elements:", L[-3:])
```

```python
average = sum(L) / len(L)

greater_than_average = [x for x in L if x > average]

print("\nAverage of L:", average)
print("Elements greater than average:", greater_than_average)
```

## RANDOM NUMBERS

```python
import random
import math

# Use the same roll number as seed
random.seed(int(roll_no))

# i. Generate 100 random numbers from 100 to 900
random_numbers = [random.randint(100, 900) for _ in range(100)]

print("\n================================")
print("Q3 - RANDOM NUMBERS")
print("================================")

print("\nRandom numbers:")
print(random_numbers)
```

```python
odd_numbers = [x for x in random_numbers if x % 2 != 0]

print("\nOdd numbers:")
print(odd_numbers)

print("Count of odd numbers:", len(odd_numbers))
```

```python
even_numbers = [x for x in random_numbers if x % 2 == 0]

print("\nEven numbers:")
print(even_numbers)

print("Count of even numbers:", len(even_numbers))


# Function to check prime number

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False

    return True
```

```python
prime_numbers = [x for x in random_numbers if is_prime(x)]

print("\nPrime numbers:")
print(prime_numbers)

print("Count of prime numbers:", len(prime_numbers))
```

```python
frequency = {}

for number in random_numbers:
    frequency[number] = frequency.get(number, 0) + 1

most_frequent = max(frequency, key=frequency.get)

print("\nMost frequently occurring number:", most_frequent)
print("Number of times it occurs:", frequency[most_frequent])
```

## SETS

```python
digits = [int(digit) for digit in roll_no]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("\n================================")
print("Q4 - SETS")
print("================================")

print("\nSet A:")
print(A)

print("\nSet B:")
print(B)
```

```python
union_set = A.union(B)

print("\nUnion of A and B:")
print(union_set)

```

```python
intersection_set = A.intersection(B)

print("\nIntersection of A and B:")
print(intersection_set)
```

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("\nA - B:")
print(A_minus_B)

print("\nB - A:")
print(B_minus_A)
```

```python
symmetric_diff = A.symmetric_difference(B)

print("\nSymmetric Difference:")
print(symmetric_diff)
```

```python
print("\nIs A a subset of B?")
print(A.issubset(B))

print("\nIs B a superset of A?")
print(B.issuperset(A))
```

```python
X = int(input("\nEnter a value X to remove from set A: "))

A.discard(X)

print("\nSet A after discard:")
print(A)
```
